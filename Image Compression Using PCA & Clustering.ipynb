```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import mean_squared_error
from skimage import io, color
```

```python
image = io.imread('your_image.jpg')  # Or use a grayscale image
image = image / 255.0  # Normalize pixel values between 0 and 1
original_shape = image.shape

# Reshape (H, W, 3) into (H*W, 3) for RGB or (H*W, 1) for grayscale
pixels = image.reshape(-1, 3)
```

```python
pca = PCA(n_components=2)  # Try different values: 1, 2, 3
pixels_pca = pca.fit_transform(pixels)

# Reconstruct to original color space
pixels_reconstructed = pca.inverse_transform(pixels_pca)
image_reconstructed_pca = pixels_reconstructed.reshape(original_shape)
```

```python
kmeans = KMeans(n_clusters=16, random_state=42)  # Try 8, 16, 32
clusters = kmeans.fit_predict(pixels)
compressed_colors = kmeans.cluster_centers_[clusters]
image_reconstructed_kmeans = compressed_colors.reshape(original_shape)
```

```python
# Apply PCA first
pca = PCA(n_components=2)
pixels_pca = pca.fit_transform(pixels)

# Then cluster PCA-compressed data
kmeans = KMeans(n_clusters=16)
clusters = kmeans.fit_predict(pixels_pca)
compressed_pca_kmeans = kmeans.cluster_centers_[clusters]
reconstructed = pca.inverse_transform(compressed_pca_kmeans)
image_reconstructed_combined = reconstructed.reshape(original_shape)
```

```python
mse = mean_squared_error(pixels, pixels_reconstructed)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(image)
axes[0].set_title("Original")

axes[1].imshow(image_reconstructed_pca)
axes[1].set_title("PCA Compression")

axes[2].imshow(image_reconstructed_kmeans)
axes[2].set_title("K-Means Compression")

for ax in axes:
    ax.axis('off')
plt.show()
```
